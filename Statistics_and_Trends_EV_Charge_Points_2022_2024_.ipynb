```python

%pip install -q odfpy

import pandas as pd
from pathlib import Path

filename = next(
    name for name in uploaded
    if name.lower().endswith(".ods")
)

sheets = pd.read_excel(
    filename,
    sheet_name=None,
    header=None,
    engine="odf"
)

for year, data in sheets.items():
    print(f"\n{'=' * 20} {year} {'=' * 20}")
    print("Shape:", data.shape)
    display(data.head(15))
```

```python

for sheet_name, data in sheets.items():
    print(f"\n{'=' * 15} SHEET: {sheet_name} {'=' * 15}")

    for idx, row in data.iterrows():
        values = [
            str(value).strip()
            for value in row.tolist()
            if pd.notna(value) and str(value).strip()
        ]

        if values:
            print(f"Row {idx}: {' | '.join(values)}")
```

```python

for sheet_name, data in sheets.items():
    print(f"\nSheet: {sheet_name}")

    text_values = set()

    for value in data.to_numpy().flatten():
        if pd.notna(value) and isinstance(value, str):
            value = value.strip()
            if value and not value.replace(",", "").replace(".", "").isdigit():
                text_values.add(value)

    print("Text labels and non-numeric entries:")
    print(sorted(text_values)[:60])
```

```python

import numpy as np
import pandas as pd

records = []

for year, data in sheets.items():
    # In each sheet, the percentage section's region rows
    # run from row index 44 through 52.
    for row_index in range(44, 53):
        region = data.iloc[row_index, 0]
        value = data.iloc[row_index, 10]

        # Skip empty labels and treat suppressed "u" values
        # as missing, never as zero.
        if pd.isna(region) or str(region).strip() == "":
            continue

        numeric_value = pd.to_numeric(value, errors="coerce")

        records.append({
            "year": int(year),
            "region": str(region).strip(),
            "charge_point_presence_pct": numeric_value
        })

df_clean = pd.DataFrame(records)

# Remove rows where the required percentage is unavailable
df_clean = df_clean.dropna(
    subset=["charge_point_presence_pct"]
).reset_index(drop=True)

df_clean["year"] = df_clean["year"].astype(int)

print("Cleaned dataset shape:", df_clean.shape)
print("\nRecords by year:")
print(df_clean.groupby("year").size())

print("\nCleaned data preview:")
display(df_clean.head(15))

print("\nMissing values:")
display(df_clean.isnull().sum())
```

```python

print("Years available:", sorted(df_clean["year"].unique()))
print("Regions available:", sorted(df_clean["region"].unique()))

print("\nSummary of charge-point presence (%):")
display(df_clean["charge_point_presence_pct"].describe().round(2))

print("\nRegional data by year:")
display(
    df_clean.pivot(
        index="region",
        columns="year",
        values="charge_point_presence_pct"
    ).round(2)
)
```

```python

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(11, 6))

sns.lineplot(
    data=df_clean,
    x="year",
    y="charge_point_presence_pct",
    hue="region",
    marker="o"
)

plt.title("EV Charge-Point Presence in Dwellings by Region (2022–2024)")
plt.xlabel("Year")
plt.ylabel("Dwellings with Charge-Point Presence (%)")
plt.xticks([2022, 2023, 2024])
plt.legend(title="Region", bbox_to_anchor=(1.02, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd
from scipy.stats import skew, kurtosis

# Select the percentage variable
values = df_clean["charge_point_presence_pct"].dropna()

# Calculate the four statistical moments
moments = pd.DataFrame({
    "Statistical Moment": [
        "Mean",
        "Variance",
        "Skewness",
        "Kurtosis (excess)"
    ],
    "Value": [
        values.mean(),
        values.var(),
        skew(values, bias=False) if len(values) > 2 else np.nan,
        kurtosis(values, fisher=True, bias=False) if len(values) > 3 else np.nan
    ]
})

display(moments.round(3))
```

```python

def calculate_moments(series):
    series = series.dropna()

    return pd.Series({
        "Mean": series.mean(),
        "Variance": series.var(),
        "Skewness": (
            skew(series, bias=False) if len(series) > 2 else np.nan
        ),
        "Kurtosis (excess)": (
            kurtosis(series, fisher=True, bias=False)
            if len(series) > 3 else np.nan
        )
    })

moments_by_year = (
    df_clean.groupby("year")["charge_point_presence_pct"]
    .apply(calculate_moments)
    .unstack()
)

display(moments_by_year.round(3))
```

```python

# Average charge-point presence across regions for each year
year_summary = (
    df_clean.groupby("year")["charge_point_presence_pct"]
    .agg(["mean", "median", "min", "max", "std", "count"])
    .round(3)
)

year_summary.columns = [
    "Mean (%)", "Median (%)", "Minimum (%)",
    "Maximum (%)", "Standard deviation", "Number of regions"
]

display(year_summary)
```

```python

# Compare regional percentages between the first and last years
regional_change = (
    df_clean.pivot(
        index="region",
        columns="year",
        values="charge_point_presence_pct"
    )
)

print("Regional change from 2022 to 2024:")

if 2022 in regional_change.columns and 2024 in regional_change.columns:
    regional_change["Change (percentage points)"] = (
        regional_change[2024] - regional_change[2022]
    )

    display(regional_change.round(3).sort_values(
        "Change (percentage points)", ascending=False
    ))
else:
    print("Check that the dataset contains numeric year columns 2022 and 2024.")
```
